```python
import numpy as np
import pandas as pd
df=pd.read_csv('datas/data.csv')
X=df.loc[:,'f1':'f100'] # матрица признаков
y=df.loc[:,'target'] # целевая переменная
```

### *РЕШЕНИЕ*

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error
```

```python
# делим датасет на тренировочную и тестовую выборку
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.33,random_state=42)
```

```python
# Обучаем линейную регрессию на тренировочных данных и оцениваем среднеквадратическую ошибку на тестовых данных
model = Ridge(random_state=42)
model.fit(X_train,y_train)

preds_val=model.predict(X_test)

answe1 = round(mean_squared_error(y_test, preds_val),2)
answe1
```
